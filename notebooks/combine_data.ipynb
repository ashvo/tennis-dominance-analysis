```python
import pandas as pd
import glob
import os

# Only load 2000-2024 (faster, and more relevant for Big Three era analysis)
files = glob.glob('../data/raw/atp_matches_2*.csv')

# Filter out doubles and futures files
files = [f for f in files if 'doubles' not in f and 'futures' not in f]

print(f'Found {len(files)} files')
print('Files:', files)
```

```python
dfs = []
for f in sorted(files):
    year = f.split('_')[-1].replace('.csv', '')
    df_year = pd.read_csv(f)
    dfs.append(df_year)
    print(f'{year}: {len(df_year):,} matches loaded')

# Combine all into one
matches = pd.concat(dfs, ignore_index=True)
print(f'\nTotal matches: {len(matches):,}')
```

```python
import os

# Create the processed folder if it doesn't exist
os.makedirs('../data/processed', exist_ok=True)

# Save matches
matches.to_csv('../data/processed/atp_matches_clean.csv', index=False)
print('✓ Matches saved')

# Load and save grand slam data
slams = pd.read_csv('../data/raw/Mens_Tennis_Grand_Slam_Winner.csv')
slams.to_csv('../data/processed/slams_clean.csv', index=False)
print('✓ Grand slam data saved')

# Load and save players
players = pd.read_csv('../data/raw/atp_players.csv')
players['player_name'] = players['name_first'] + ' ' + players['name_last']
players.to_csv('../data/processed/players_clean.csv', index=False)
print('✓ Players saved')

print(f'\nDone! Check data/processed/')
```
